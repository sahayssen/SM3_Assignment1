```python
import pandas as pd
```

Open data

```python
data = pd.read_csv("../data/UNDER-CUSTODY-EXTRACT.CSV")
```

Check data

```python
data.head(10)
```

```python
data['FACILITY_NAME'].value_counts()
```

Clean data

eliminate minimum secruity prisons

```python
data = data[data['FACILITY_SECURITY_LEVEL'] != 'MIN']
```

limit data to most recent report date

```python

data = data[data['REPORT_DATE'] == '2026-08-01']
```

limit data to only open facilities

```python
data  = data[data['FACILITY_STATUS'] == 'OPEN']
```

```python
data
```

```python
#missing_drs = data[(data['FACILITY_NAME']=='COLLINS')|(data['FACILITY_NAME']=='CAYUGA')|(data['FACILITY_NAME']=='WENDE')|(data['FACILITY_NAME']=='FIVE POINTS')]
```

Group data set by to find the racial breakdown of selected facilities 

```python
race_counts = data.groupby(['FACILITY_NAME', 'FACILITY_SECURITY_LEVEL','RACE'])['POPULATION_IN_CUSTODY'].sum().reset_index(name='counts')
```

```python
race_counts 
```

Make a column with the totals for each prison 

```python
totals = data.groupby(['FACILITY_NAME', 'FACILITY_SECURITY_LEVEL'])['POPULATION_IN_CUSTODY'].sum().reset_index(name='counts')
```

```python
totals
```

```python
totals['RACE'] = 'TOTAL'
```

```python
totals
```

```python
totals.dtypes
```

```python
race_counts.dtypes
```

Merge totals with race count

```python
facility_merged = pd.concat([totals,race_counts], ignore_index=True)
```

```python
facility_merged
```

Add rows with the totals for each prison type to bottom of dataset

```python
security_totals = data.groupby(['FACILITY_SECURITY_LEVEL'])['POPULATION_IN_CUSTODY'].sum().reset_index(name='counts')
```

```python
security_totals
```

```python
security_totals['RACE'] = 'NONE'
```

```python
security_totals['FACILITY_NAME'] = 'TOTAL'
```

```python
security_totals
```

Merge security totals with larger count

```python
facility_merged = pd.concat([facility_merged,security_totals], ignore_index=True)
```

```python
facility_merged 
```

export

```python
facility_merged.to_csv('../output/facility_merged.csv', index=False)
```
